# **A Famous Syllogism**

```python
ASSUMPTIONS = """
% A1: All Greeks are men.
all x (Greek(x) -> Man(x)).

% A2: All men are mortal.
all x (Man(x) -> Mortal(x)).
"""

GOAL = """
% Goal: All Greeks are mortal.
all x(Greek(x) -> Mortal(x)).
"""
```

```python
pyprover9.prove(ASSUMPTIONS,GOAL)
```

# **Homework is Fun**

```python
ASSUMPTIONS = """
% A1 No homework is fun.
all x (Homework(x) -> -Fun(x)).

% A2 Some reading is homework.
exists x (Reading(x) & Homework(x)).
"""

GOAL = """
% Goal: Some reading is not fun.
exists x (Reading(x) & -Fun(x)).
"""
```

```python
pyprover9.prove(ASSUMPTIONS,GOAL)
```

# **Greedy Rabbits**

```python
ASSUMPTIONS = """
% A1	No old rabbits are greedy.
all x (Rabbit(x) & Old(x) -> -Greedy(x)).

% A2  All black rabbits are greedy.
all x ((Rabbit(x) & Black(x)) -> Greedy(x)).
"""

GOAL = """
% Goal  No old rabbits are black.
all x ((Rabbit(x) & Old(x)) -> -Black(x)).
"""
```

```python
pyprover9.prove(ASSUMPTIONS,GOAL)
```

# **Young Rabbits**

```python
ASSUMPTIONS = """
% A1	Young rabbits are greedy.
all x ((Rabbit(x) & Young(x)) -> Greedy(x)).

% A2  Greedy animals are fat.
all x ((Animal(x) & Greedy(x)) -> Fat(x)).

% A3  Rabbits are animals.
all x (Rabbit(x) -> Animal(x)).
"""

GOAL = """
% Goal  All young rabbits are fat.
all x ((Rabbit(x) & Young(x)) -> Fat(x)).
"""
```

```python
pyprover9.prove(ASSUMPTIONS,GOAL)
```

# **An Invalid Syllogism**

```python
ASSUMPTIONS = """
% A1 Some rhombuses are equilateral.
exists x (Rhombus(x) & Equilateral(x)).

% A2 Some equilaterals are triangular.
exists x (Equilateral(x) & Triangular(x)).
"""

GOAL = """
% Goal Some rhombuses are triangular.
exists x (Rhombus(x) & Triangular(x)).
"""

pyprover9.prove(ASSUMPTIONS, GOAL)
```

# **Porky**

```python
ASSUMPTIONS = """
% A1 Porky is a pig.
Pig(Porky).

% A2 All pigs are animals.
all x (Pig(x) -> Animal(x)).

% A3 All pigs are fat.
all x (Pig(x) -> Fat(x)).

% A4 No fat animals run fast.
all x ((Animal(x) & Fat(x)) -> -Fast(x)).
"""

GOAL = """
% Porky does not run fast.
-Fast(Porky).
"""
```

```python
pyprover9.prove(ASSUMPTIONS,GOAL)
```

# **Jabberwock**

```python
ASSUMPTIONS = """
% A1 The Jabberwock is a creature with eyes of flame.
Creature(Jabberwock) & FlamyEyes(Jabberwock).

% A2 All creatures with eyes of flame whiffle and burble.
all x ((FlamyEyes(x) & Creature(x)) -> (Whiffle(x) & Burble(x))).

% A3 All creatures that whiffle or burble have claws.
all x (((Whiffle(x) | Burble(x)) & Creature(x)) -> Claws(x)).
"""

GOAL = """
% The Jabberwock has claws.
Claws(Jabberwock).
"""
```

```python
pyprover9.prove(ASSUMPTIONS,GOAL)
```

# **Balloonig Pigs**

```python
ASSUMPTIONS = """
% A1
all x ((-Dance_on_Tightropes(x) & -Eat_Penny_Buns(x)) -> Old(x)).

% A2
all x ((Pig(x) & Giddy(x)) -> Respected(x)).

% A3
all x ((Wise(x) & Balloonist(x)) -> Takes_Umbrella(x)).

% A4
all x ((Ridiculous(x) & Eat_Penny_Buns(x)) -> -Lunch_in_Public(x)).

% A5
all x ((Young(x) & Balloonist(x)) -> Giddy(x)).

% A6
all x ((Fat(x) & Ridiculous(x) & -Dance_on_Tightropes(x)) -> Lunch_in_Public(x)).

% A7
all x ((Wise(x) & Giddy(x)) -> -Dance_on_Tightropes(x)).

% A8
all x ((Pig(x) & Takes_Umbrella(x)) -> Ridiculous(x)).

% A9
all x ((-Dance_on_Tightropes(x) & Respected(x)) -> Fat(x)).

% A10
all x -(Young(x) & Old(x)).
"""

GOAL = """
all x ((Wise(x) & Young(x) & Pig(x)) -> -Balloonist(x)).
"""

pyprover9.prove(ASSUMPTIONS, GOAL)
```
