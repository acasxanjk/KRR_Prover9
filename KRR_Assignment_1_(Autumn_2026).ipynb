# **The Commuter**

```python
%pip install -q --upgrade pyprover9
import pyprover9
```

```python
ASSUMPTIONS = """
% A1
(Windy -> Raining) & (Raining -> -GoBySkateboard).

% A2
GoToUniversity -> -SplashedByCars.

% A3
(Outside & -WearHelmet & Raining) -> HairWet.

% A4
GoToUniversity -> (GoBySkateboard | GoByTaxi | GoOnFoot).

% A5
WearHelmet <-> GoBySkateboard.

% A6
HearWell -> (-WearHelmet & -HairWet).

% A7
(Raining & Outside & -HearWell) -> SplashedByCars.

% A8
(GoToUniversity & (GoBySkateboard | GoOnFoot)) -> Outside.
"""

GOAL = """
(Windy & GoToUniversity) -> GoByTaxi.
"""

pyprover9.prove(ASSUMPTIONS, GOAL)
```

# **Sunny Farm**

```python
ASSUMPTIONS = """
% A1: At most one shy animal lives at Sunny Farm.
all x all y (
    (Animal(x) & Shy(x) & LivesAt(x, sunny_farm) &
     Animal(y) & Shy(y) & LivesAt(y, sunny_farm))
    -> x = y
).

% A2: Animals liked by another animal are happy.
all x (
    (Animal(x) &
     exists y (Animal(y) & -(y = x) & Likes(y, x)))
    -> Happy(x)
).

% A3: No carnivores live at Sunny Farm.
all x (Carnivore(x) -> -LivesAt(x, sunny_farm)).

% A4: Poppy and Cutie are two distinct rabbits at Sunny Farm.
Rabbit(poppy) & Rabbit(cutie) &
LivesAt(poppy, sunny_farm) & LivesAt(cutie, sunny_farm) &
-(poppy = cutie).

% A5: At least three distinct animals live at Sunny Farm.
exists x exists y exists z (
    Animal(x) & Animal(y) & Animal(z) &
    LivesAt(x, sunny_farm) &
    LivesAt(y, sunny_farm) &
    LivesAt(z, sunny_farm) &
    -(x = y) & -(x = z) & -(y = z)
).

% A6: Rabbits are non-carnivorous animals.
all x (Rabbit(x) -> (Animal(x) & -Carnivore(x))).

% A7: Every non-fox animal at Sunny Farm is gentle or shy.
all x (
    (Animal(x) & LivesAt(x, sunny_farm) & -Fox(x))
    -> (Gentle(x) | Shy(x))
).

% A8: Definition of a carnivore.
all x (
    Carnivore(x) <->
    (Animal(x) &
     exists y (Animal(y) & -(x = y) & LikesToEat(x, y)))
).

% A9: Every gentle animal likes every other gentle animal.
all x all y (
    (Animal(x) & Gentle(x) &
     Animal(y) & Gentle(y) & -(x = y))
    -> Likes(x, y)
).

% A10: All foxes like to eat every rabbit.
all x all y (
    (Fox(x) & Rabbit(y)) -> LikesToEat(x, y)
).
"""

GOAL = """
exists x (
    Rabbit(x) & Happy(x) & LivesAt(x, sunny_farm)
).
"""

pyprover9.prove(ASSUMPTIONS, GOAL)
```
