# **The Stolen Diamond**

```python
%pip install -q --upgrade pyprover9
import pyprover9
```

```python
ASSUMPTIONS = """
% A1	The diamond has been stolen!
DiamondStolen.

% A2	If the diamond has been stolen, a thief got into the house.
DiamondStolen -> ThiefInHouse.

% A3	If a thief got in, either the door was open or the window was smashed.
(ThiefInHouse -> (DoorOpen | WindowSmashed)).

% A4	The window is not smashed.
-WindowSmashed.
"""

GOAL = """
% Goal	The door was open.
DoorOpen.
"""
```

```python
pyprover9.prove(ASSUMPTIONS,GOAL)
```

# **The Lost Key**

```python
ASSUMPTIONS = """
% A1   If Alan did not lock the door he either forgot or lost the key.
-AlanLockedDoor -> (AlanForgotKey | AlanLostKey).

% A2   If he lost the key he will be in trouble.
AlanLostKey -> AlanInTrouble.

% A3   If he has a good memory he did not forget.
AlanGoodMemory -> -AlanForgotKey.

% A4   Alan has a good memory.
AlanGoodMemory.
"""

GOAL = """
% Goal   Either Alan locked the door or he will be in trouble.
AlanLockedDoor | AlanInTrouble.
"""
```

```python
pyprover9.prove(ASSUMPTIONS,GOAL)
```

# **Rembrandt's Picture**

```python
ASSUMPTIONS = """
% A1.  Either the picture is valuable or it is a fake.
PictureIsValuable | PictureIsFake.

% A2.  If it is valuable it is by Rembrandt.
PictureIsValuable -> PictureByRembrandt.

% A3.  If it is a fake there will be a police investigation.
PictureIsFake -> PoliceInvestigation.
"""

GOAL = """
% Goal   Either the picture is by Rembrandt or there will be a police investigation.
PictureByRembrandt | PoliceInvestigation.
"""
```

```python
pyprover9.prove(ASSUMPTIONS,GOAL)
```

# **Sunday is a Holiday**

```python
ASSUMPTIONS = """
% A1.  Sunday is a Holiday.
Sunday -> Holiday.

% A2.  Saturday and Sunday is the weekend.
(Saturday|Sunday) <-> Weekend.

% A3.  On holidays the University is closed.
Holiday -> UniversityClosed.

% A4.  It is the weekend.
Weekend.

% A5.  It is not Saturday.
-Saturday.
"""

GOAL = """
% Goal   The University is closed.
UniversityClosed.
"""
```

```python
pyprover9.prove(ASSUMPTIONS,GOAL)
```

# **Tarquin's Wisdom**

```python
ASSUMPTIONS = """
% A1.  f Tarquin learns philosophy then he acquires wisdom and if he learns science he will advance human knowledge.
(TarquinLearnsPhilosophy -> TarquinAcquiresWisdom) & (TarquinLearnsScience -> TarquinAdvancesKnowledge).

% A2.  If he acquires wisdom, Tarquin will not die in misery.
TarquinAcquiresWisdom -> -TarquinDiesMisery.

% A3.  If he advances human knowledge, his aspirations will be satisfied.
TarquinAdvancesKnowledge -> TarquinSatisfiesAspirations.

% A4.  If Tarquin's aspirations are not satisfied he will die in misery.
-TarquinSatisfiesAspirations -> TarquinDiesMisery.

% A5.  Alas, Tarquin's aspirations will never be satisfied.
-TarquinSatisfiesAspirations.
"""

GOAL = """
% Goal   Thus, he learns neither philosophy nor science.
-TarquinLearnsPhilosophy & -TarquinLearnsScience.
"""
```

```python
pyprover9.prove(ASSUMPTIONS,GOAL)
```

# **Dry Thunder**

```python
ASSUMPTIONS = """
% A1.   A Dry Thunderstorm is defined by the occurence of lightning without rain.
(Lightning & -Raining) <-> DryThunder.

% A2.  	There is a dry thunderstorm.
DryThunder.

% A3.   If there is lightning then it is either raining or its hot.
Lightning -> (Raining | Hot).
"""

GOAL = """
% Goal   It is hot.
Hot.
"""
```

```python
pyprover9.prove(ASSUMPTIONS,GOAL)
```

# **Angry Dragons**

```python
ASSUMPTIONS = """
% A1.  If the dragon destroys the village, the villagers will be homeless unless they take the King's Castle.
DragonDestroysVillage -> (VillagersHomeless <-> -VillagersTakeCastle).

% A2.  If the King is mean then the villagers will take the castle.
KingMean -> VillagersTakeCastle.

% A3.  If the King is not mean then the dragon won't destroy the village.
-KingMean -> -DragonDestroysVillage.

% A4.  The dragon destroys the village.
DragonDestroysVillage.
"""

GOAL = """
% Goal   The villagers are not homeless.
-VillagersHomeless.
"""
```

```python
pyprover9.prove(ASSUMPTIONS,GOAL)
```
